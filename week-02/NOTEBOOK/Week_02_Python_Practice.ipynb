List — ordered, changeable collection

A list stores multiple items in one variable. You can add, remove, or change items. Items are in order and can repeat.

QUESTION:
Product Inventory Search;


```python
inventory = [
    ["Laptop", 15],
    ["Mouse", 5],
    ["Keyboard", 0]
]

search = input("Product: ")
found = False

for item in inventory:
    if item[0].lower() == search.lower():
        found = True

        if item[1] > 0:
            print("Available")
        else:
            print("Out of Stock")

if not found:
    print("Product not found")
```

2. Dictionary
Dictionary — key-value pairs

A dictionary stores data as key:value pairs. Like a real dictionary — you look up a word (key) to get its meaning (value).

Question : Student Performance Dashboard
Store marks and calculate:
Average
Highest marks
Students above average

```python
students={
"Ali":87,
"Sara":95,
"Abeer":78,
"Ahmed":82,
"Saba":90
}

avg=sum(students.values())/len(students)

top=max(students,key=students.get)

print("Average:",avg)

print("Topper:",top)

for s,m in students.items():

    if m>avg:

        print(s)
```

3. Tuples
Tuple — ordered, unchangeable collection
Like a list but fixed — you cannot change its values after creation. Use for data that should stay constant.


Question : Monthly Revenue Analysis


```python
sales=(10000,15000,12000,17000)

print("Total:",sum(sales))

print("Avg:",sum(sales)/len(sales))

print("High:",max(sales))
```

4. Functions
Function — reusable block of code
A function runs a task whenever you call it. Write once, use many times. Use def to define it.

Question : Student Grade Function

```python
def grade(score):

    if score>=90:
        return "A"

    elif score>=80:
        return "B"

    elif score>=70:
        return "C"

    return "Fail"

marks=[92,81,76,40]

for m in marks:

    print(grade(m))
```

6. Conditional Statements
Conditional statement — decision making
Use if / elif / else to run different code based on a condition. Like asking "is this true? if yes do this, if no do that."

Question : Login Authentication

```python
username = "admin"
psw = "123"

attempt = 0

while attempt < 3:
    u = input("Enter username: ")
    p = input("Enter password: ")

    if u == username and p == psw:
        print("Success")
        break

    attempt += 1

else:
    print("Account Locked")
```

8. CSV Reading
CSV files store data in rows and columns separated by commas. csv.DictReader reads each row as a dictionary using column headers as keys.

```python
import csv

total = 0

try:
    with open("sales.csv", "r") as file:
        reader = csv.DictReader(file)
        for row in reader:
            total += int(row["Price"])
    print(total)

except FileNotFoundError:
    print("Error: File not found.")
except ValueError:
    print("Error: Invalid price value found.")
```

9. Exception Handling
try / except — handle errors gracefully

When code might fail (file missing, wrong input), wrap it in try. If an error happens, except catches it so the program doesn't crash.

Common Exceptions:
ZeroDivisionError
ValueError
FileNotFoundError
IndexError
KeyError

```python
try:

    a=int(input())

    b=int(input())

    print(a/b)

except ZeroDivisionError:

    print("Cannot divide")

except ValueError:

    print("Numbers only")
```

Create a Python program that reads a CSV file and displays: 
Total number of rows 
Total number of columns 
Column names 
Missing values in each column 
Basic statistical summary 

```python

import pandas as pd

def analyze_csv():
    try:
        df = pd.read_csv("Student_dataset.csv")
        print("Dataset Information")
        print(f"Total Rows: {df.shape[0]}")
        print(f"Total Columns: {df.shape[1]}")
        print()
        print("Coloumn names")
        print(list(df.columns))
        print()
        print("Missing value ")
        print(df.isnull().sum())
        print()
        print("Stastical summary")
        print(df.describe(include='all'))
    except FileNotFoundError:
        print("Error: CSV file not found.")
    except pd.errors.EmptyDataError:
        print("Error: CSV file is empty.")
    except Exception as e:
        print(f"An error occurred: {e}")


analyze_csv()
```
